## Data collection

```python
df = pd.read_csv('DataTime_export.csv')  # 47 features
df = df.rename(columns={'VD0000':'impl_spd2',  # Main fan speed [rps]
                        'VD0008':'pump_spd',  # Peristaltic pump rotation speed [rps]
                        'VD0012':'I',  # Current [A]
                        'VD0016':'V',  # Voltage [V]
                        'VD0020':'W',  # Power [Wt]
                        'VD0028':'Q_CH4',  # CH4 flow rate [m3/h]
                        'VD0032':'impl_spd1',  # Cooling fan speed [rps]
                        'VD0036':'Q_CH4_N2',  #CH4/N2 flow rate [m3/h]
                        'VD0040':'P_NG',  # Differential natural gas pressure [bar]
                        'VD44':'O2',  # Oxygen concentration at the burner inlet [%]
                        'VD208':'T3',  # Temperature at the bottom of the burner [°C]
                        'VD212':'T4',  # Temperature at the top of the burner [°C]
                        'VD216':'T5',  # temperature at the inlet of the reformer [°C]
                        'VD224':'T7',  # SOFC exhaust gases temperature [°C]
                        'VD232':'T9',  # heat exchanger temperature [°C]
                        'VD244':'T12',  # water temperature for steam reforming [°C]
                        'VD260':'T16',  # temperature at the SOFC left front point [°C]
                        'VD264':'T17',  # temperature at the SOFC right rear point [°C]
                        'VD272':'T19',  # air temperature at the SOFC inlet [°C]
                        'VD276':'T20',  # hydrogen temperature at the SOFC inlet [°C]
                        'VD280':'T21',  # air temperature at the SOFC outlet [°C]
                        'VD284':'T22',  # hydrogen temperature at the SOFC outlet [°C]
                        'VD288':'T23',  # temperature at the top rear point of the reformer [°C]
                        'VD292':'T24',  # temperature at the bottom rear point of the reformer [°C]
                        'VD296':'T25',  # temperature at the top left point of the reformer [°C]
                        'VD300':'T26',  # temperature at the bottom left point of the reformer [°C]
                        'VD304':'T27',  # temperature at the top right point of the reformer [°C]
                        'VD308':'T28',  # temperature at the bottom right point of the reformer [°C]
                        'VD316':'T30',  # cooling water temperature [°C]
                        'VD320':'T31'})  # water tank temperature [°C]

df.MCGS_TIME = pd.to_datetime(df.MCGS_TIME)
df = df.set_index('MCGS_TIME')
df = df.drop(columns=['MCGS_TIMEMS', 'VD48', 'VD0004', 'VD0024'])  # drop 4 useless features
df = df[df.columns[(abs(df.mean()-3000)>0.1)]]  # filtering 12 incorrect sensor readings
df = df[df.T17 != 3000]  # remove 92 lines with incorrect T17-value
df = df[df.V != 0]  # remove 18324 lines with zero target feature

# move the voltage to the end position (for correlation matrix)
V = df.V
df = df.drop('V', axis=1)
df['V'] = V

#df.to_excel(f'DATA.xlsx')

print(df.info())
df.iloc[3430:3440, :]
```

## Data preprocessing

```python
X = df.drop(columns='V').copy()  # input feature
y = df.V.copy()  # target feature (Voltage [V])

# Normalization of input feature general set 
scaler = MinMaxScaler()
X_norm = scaler.fit_transform(X)
X_norm = pd.DataFrame(X_norm, columns=X.columns)

# Spliting normalization data into train and test sets (for MLP Analisys)
X_train_norm, X_test_norm, y_train_norm, y_test_norm = train_test_split(X_norm, y, test_size=0.2, random_state=random_state)

# Spliting legacy data into train and test sets (for GB & RF Analisys)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=random_state)
```

## Feature important

```python
n = 15  # base component count
```

### CatBoost features importance (MDI)

```python
selector = catboost.CatBoostRegressor(random_state=random_state, silent=True)
selector.fit(X, y)

feature_importances = pd.DataFrame({'importance' : selector.feature_importances_}, 
                                   index=X.columns).sort_values(by='importance', ascending=False)
```

```python
fontsize = 18

n_feature_importances = feature_importances.iloc[:n, :].copy()  # get n-strongest features
MDI_columns = n_feature_importances.sort_values(by='importance', ascending=False).index

plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['font.size'] = fontsize

plt.figure(figsize=(n, 4.5))

for bar in plt.bar(range(len(n_feature_importances.index)), n_feature_importances.importance, 
                   align='center', width=0.8, color='#ff0051'):
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval, round(yval, 3), ha='center', va='bottom')
        
plt.xticks(range(len(n_feature_importances.index)), n_feature_importances.index, rotation=45)
plt.ylabel('Вес признака, %', fontsize=fontsize)
plt.ylim(0, ((feature_importances.importance.max() // 5) + 1) * 5)
plt.title('Ценность признаков в контексте CatBoost-регрессора с дефолтным MDI-оценщиком', fontsize=fontsize)
plt.grid(axis='x', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.savefig('pictures/GB_standard_feature_important.png', dpi = 300, transparent = True)
plt.show()
```

### Permutation Importance (PI)

```python
feature_importances = pd.DataFrame(
    {'importance' : permutation_importance(selector, X, y, n_repeats=5, 
                                           random_state=random_state, n_jobs=-1).importances_mean},
    index=X.columns).sort_values(by='importance', ascending=False)

feature_importances.importance = feature_importances.importance * 100
```

```python
n_feature_importances = feature_importances.iloc[:n, :].copy()  # get n-strongest features
PI_columns = n_feature_importances.sort_values(by='importance', ascending=False).index

plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['font.size'] = fontsize

plt.figure(figsize=(n, 4.5))

for bar in plt.bar(range(len(n_feature_importances.index)), n_feature_importances.importance, 
                   align='center', width=0.8, color='#ff0051'):
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval, round(yval, 3), ha='center', va='bottom')

plt.xticks(range(len(n_feature_importances.index)), n_feature_importances.index, rotation=45)
plt.ylim(0, ((feature_importances.importance.max() // 5) + 1) * 5)
plt.ylabel('Вес признака, %', fontsize=fontsize)
plt.title('Ценность признаков в контексте CatBoost-регрессора с PI-оценщиком', fontsize=fontsize)
plt.grid(axis='x', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.savefig('pictures/GB_Permutation_Importance.png', dpi = 300, transparent = True)
plt.show()
```

### SHapley Additive exPlanations (SHAP)

```python
explainer = shap.Explainer(selector)
shap_importances = explainer(X)

shap_values = np.abs(shap_importances.values).mean(0)
feature_names = shap_importances.feature_names

feature_importances = pd.DataFrame({
                        'feature': feature_names,
                        'importance': shap_values
                    })

feature_importances = feature_importances.sort_values(by='importance', ascending=False).head(n)
SHAP_columns = list(feature_importances.feature)

plt.figure(figsize=(n, 4.5))

for bar in plt.bar(feature_importances.feature, feature_importances.importance, 
                   align='center', width=0.8, color='#ff0051'):
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval, round(yval, 3), ha='center', va='bottom')

plt.title('Ценность признаков в контексте CatBoost-регрессора с SHAP-оценщиком', fontsize=fontsize)
plt.ylabel('среднее(|SHAP-значение|)', fontsize=fontsize)
plt.ylim(0, ((feature_importances.importance.max() // 1) + 1))

plt.xticks(rotation=45) 
plt.grid(axis='x', linestyle='--', alpha=0.3)
plt.tight_layout()

plt.savefig('pictures/GB_SHAP_Importance.png', dpi=300, transparent=True)
plt.show()
```

## Gradient Boosting

```python
# CatBoost GridSearchCV option

n_splits=5

param_grid={'max_depth': range(3, 6),
            'learning_rate': [0.001, 0.1, 0.2, 0.3],
            'n_estimators': [500, 1000, 1500],
           }
```

### default (full set)

```python
%%time

start_time = time.time()
GB_grid = GridSearchCV(estimator=catboost.CatBoostRegressor(random_state=random_state, silent=True),
                       param_grid=param_grid, 
                       verbose=1,
                       #cv=ShuffleSplit(n_splits=n_splits, random_state=random_state),
                       cv=n_splits,
                       scoring='neg_mean_absolute_percentage_error'
                       )

GB_grid.fit(X_train, y_train)
GB_best = GB_grid.best_estimator_
y_pred = GB_best.predict(X_test)

elapsed_time = time.time() - start_time

print("Best CatBoostRegressor-params:", GB_grid.best_params_)
```

```python
search_result.loc['CatBoost default'] = (
                                         r2_score(y_pred, y_test),
                                         MSE(y_pred, y_test),
                                         MAE(y_pred, y_test),
                                         MAPE(y_pred, y_test),
                                         elapsed_time/(len(ParameterGrid(param_grid))*n_splits)
                                        )
print('CatBoost default',
      f'r2_score: {r2_score(y_pred, y_test):.4}',
      f'MSE: {MSE(y_pred, y_test):.4}',
      f'MAE: {MAE(y_pred, y_test):.4}',
      f'MAPE: {MAPE(y_pred, y_test):.2%}', sep='\n')
```

### PCA

```python
%%time

start_time = time.time()
df_PCA_GB_search = pd.DataFrame(columns=['r2_score', 'MSE', 'MAE', 'MAPE'])

for n in trange(5, X_train_norm.shape[1] + 1, 2, desc=f"components progress"):
    pca = PCA(n_components=n)
    pca.fit(X_norm)
    X_dec = pca.fit_transform(X_norm)
    X_dec = pd.DataFrame(X_dec, columns=[f'component{i}' for i in range(n)])

    X_dec_train2, X_dec_test2, y_train, y_test = train_test_split(X_dec, y, test_size=0.2, random_state=random_state)
    
    GB_grid = GridSearchCV(estimator=catboost.CatBoostRegressor(random_state=random_state, silent=True),
                           param_grid=param_grid, 
                           verbose=0,
                           #cv=ShuffleSplit(n_splits=n_splits, random_state=random_state),
                           cv=n_splits
                           )

    GB_grid.fit(X_dec_train2, y_train)
    print(f"Best CatBoostRegressor-params for {n} components", GB_grid.best_params_)
    
    GB_best = GB_grid.best_estimator_
    y_pred = GB_best.predict(X_dec_test2)
    
    df_PCA_GB_search.loc[n] = (r2_score(y_pred, y_test),
                               MSE(y_pred, y_test), 
                               MAE(y_pred, y_test),
                               MAPE(y_pred, y_test))

elapsed_time = time.time() - start_time
```

```python
n = df_PCA_GB_search.MAPE.idxmin()  # get optimal components number
search_result.loc[f'CatBoost+PCA({n})'] = (df_PCA_GB_search.loc[n].r2_score,
                                           df_PCA_GB_search.loc[n].MSE,
                                           df_PCA_GB_search.loc[n].MAE,
                                           df_PCA_GB_search.loc[n].MAPE,
                                           elapsed_time/(len(ParameterGrid(param_grid))*
                                                    n_splits*(len(range(5, X_train.shape[1] + 1, 2)))))
print(f'PCA {n} components',
      f'r2_score: {df_PCA_GB_search.loc[n].r2_score:.4}',
      f'MSE: {df_PCA_GB_search.loc[n].MSE:.4}',
      f'MAE: {df_PCA_GB_search.loc[n].MAE:.4}',
      f'MAPE: {df_PCA_GB_search.loc[n].MAPE:.2%}', sep='\n')
```

```python
# graph MLP
def MPL_graph(model, feature_names=None, target_name="Target"):
    dot = graphviz.Digraph(comment='MLP Architecture', graph_attr={'rankdir': 'LR', 'splines': 'line'})
    
    # layers: input + hidden + output
    layer_sizes = [model.n_features_in_] + list(model.hidden_layer_sizes) + [model.n_outputs_]
    
    # Nodes for each layer
    for i, size in enumerate(layer_sizes):
        with dot.subgraph(name=f'cluster_{i}') as c:
            c.attr(label=f'Layer {i} ({"Input" if i==0 else "Hidden" if i<len(layer_sizes)-1 else "Output"})')
            #c.attr(label=f'Слой {i} ({"Входной" if i==0 else "Скрытый" if i<len(layer_sizes)-1 else "Выходной"})')
            for j in range(size):
                node_id = f'l{i}n{j}'
                if i == 0 and feature_names is not None:
                    label = feature_names[j]
                elif i == len(layer_sizes) - 1:
                    label = target_name
                else:
                    #label = f'Узел {j}'
                    label = f'Node {j}'
                c.node(node_id, label)

    # Adding edges (weights)
    for i, weights in enumerate(model.coefs_):
        # i – is the index of the current weight layer (between layer i and i+1)
        for start_node in range(weights.shape[0]):
            for end_node in range(weights.shape[1]):
                weight_val = weights[start_node, end_node]
                # Draw only significant connections (optional) so as not to overload the graph
                if abs(weight_val) > 0.05: 
                    color = '#ff0051' if weight_val > 0 else '#008bfb'
                    width = str(abs(weight_val) * 0.25)  # The thickness of the line depends on the weight
                    dot.edge(f'l{i}n{start_node}', f'l{i+1}n{end_node}', color=color, penwidth=width)
    
    return dot
```

```python
%%time

start_time = time.time()
df_PCA_MLP_search = pd.DataFrame(columns=['r2_score', 'MSE', 'MAE', 'MAPE'])

for n in trange(5, X_train.shape[1]+1, 2, desc=f"components progress"):
    pca = PCA(n_components=n)
    pca.fit(X_norm)
    X_dec = pca.fit_transform(X_norm)
    X_dec = pd.DataFrame(X_dec, columns=[f'component{i}' for i in range(n)])

    X_dec_train, X_dec_test, y_train, y_test = train_test_split(X_dec, y, test_size=0.2, random_state=random_state)
    
    MLP_grid = GridSearchCV(estimator=MLPRegressor(random_state=random_state, 
                                                   learning_rate='invscaling',
                                                   learning_rate_init=0.055,
                                                   solver='adam',
                                                   max_iter=2000),
                            param_grid=param_grid,
                            scoring='neg_mean_absolute_percentage_error',
                            verbose=0,
                            cv=ShuffleSplit(n_splits=n_splits, random_state=random_state)
                            )
    
    MLP_grid.fit(X_dec_train, y_train)
    print(f"Best MLP-params for {n} components", MLP_grid.best_params_)
    
    MLP_best = MLP_grid.best_estimator_
    y_pred = MLP_best.predict(X_dec_test)
    
    df_PCA_MLP_search.loc[n] = (r2_score(y_pred, y_test),
                                MSE(y_pred, y_test), 
                                MAE(y_pred, y_test),
                                MAPE(y_pred, y_test))

elapsed_time = time.time() - start_time
```

```python
n = df_PCA_MLP_search.MAPE.idxmin()
search_result.loc[f'MLP+PCA({n})'] = (df_PCA_MLP_search.loc[n].r2_score,
                                        df_PCA_MLP_search.loc[n].MSE,
                                        df_PCA_MLP_search.loc[n].MAE,
                                        df_PCA_MLP_search.loc[n].MAPE,
                                        elapsed_time/(len(ParameterGrid(param_grid))*
                                                      n_splits*len(range(5, X_train.shape[1]+1, 2))))
print(f'PCA {n} components',
      f'r2_score: {df_PCA_MLP_search.loc[n].r2_score:.4}',
      f'MSE: {df_PCA_MLP_search.loc[n].MSE:.4}',
      f'MAE: {df_PCA_MLP_search.loc[n].MAE:.4}',
      f'MAPE: {df_PCA_MLP_search.loc[n].MAPE:.2%}', sep='\n')
```

```python
selector = MLPRegressor(random_state=random_state, 
                        learning_rate='invscaling',
                        learning_rate_init=0.055,
                        activation='logistic',
                        hidden_layer_sizes=(20, 20),
                        solver='adam',
                        max_iter=2000)
selector.fit(X_norm, y)
```

```python
feature_importances = pd.DataFrame(
    {'importance' : permutation_importance(selector, X_norm, y, n_repeats=5, 
                                           random_state=random_state, n_jobs=-1).importances_mean},
    index=X_norm.columns).sort_values(by='importance', ascending=False)

feature_importances.importance = feature_importances.importance * 100
```

```python
n = 15  # base component count
n_feature_importances = feature_importances.iloc[:n, :].copy()  # get n-strongest features
PI_columns = n_feature_importances.sort_values(by='importance', ascending=False).index

fontsize = 18
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['font.size'] = fontsize

plt.figure(figsize=(n, 4.5))

for bar in plt.bar(range(len(n_feature_importances.index)), n_feature_importances.importance, 
                   align='center', width=0.8, color='#ff0051'):
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval, round(yval, 3), ha='center', va='bottom')

plt.xticks(range(len(n_feature_importances.index)), n_feature_importances.index, rotation=45)
plt.ylim(0, ((feature_importances.importance.max() // 10) + 1) * 10)
plt.ylabel('Вес признака, %', fontsize=fontsize)
plt.title('Ценность признаков в контексте MLP-регрессора с PI-оценщиком', fontsize=fontsize)
plt.grid(axis='x', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.savefig('pictures/MLP_Permutation_Importance.png', dpi = 300, transparent = True)
plt.show()
```
